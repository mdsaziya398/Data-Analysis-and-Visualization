```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("All libraries imported successfully!")
```

```python
# Load dataset
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"

df = pd.read_csv(url)

# Display first 5 rows
df.head()
```

```python
df = df.drop_duplicates()

print("Duplicates removed successfully!")
```

```python
print(df.isnull().sum())
print("Final Dataset Shape:", df.shape)
```

```python
average = np.mean(df["sepal_length"])

print("Average Sepal Length:", average)
```

```python
print("Maximum Petal Length:",
      np.max(df["petal_length"]))

print("Minimum Petal Length:",
      np.min(df["petal_length"]))
```

```python
print(df["species"].value_counts())
```

```python
result = df.groupby("species").mean(numeric_only=True)

print(result)
```

```python
species_avg = df.groupby("species")["sepal_length"].mean()

plt.figure(figsize=(8,5))

plt.bar(species_avg.index, species_avg.values)

plt.title("Average Sepal Length by Species")
plt.xlabel("Flower Species")
plt.ylabel("Average Sepal Length")

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(df.index[:20],
         df["sepal_length"].iloc[:20],
         marker="o")

plt.title("Sepal Length of First 20 Flowers")
plt.xlabel("Flower Record")
plt.ylabel("Sepal Length")

plt.grid(True)
plt.show()
```

```python
species_count = df["species"].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    species_count.values,
    labels=species_count.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Distribution of Iris Flower Species")

plt.show()
```

```python
print("DATA ANALYSIS COMPLETED")

print("\nFirst 5 Records:")
display(df.head())

print("\nDataset Information:")
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

print("\nSpecies Distribution:")
print(df["species"].value_counts())

print("\nAverage Measurements:")
display(df.groupby("species").mean(numeric_only=True))

print("\nMissing Values:")
print(df.isnull().sum())
```
